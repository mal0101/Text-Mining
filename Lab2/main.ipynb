#### 1.2. Comptage de mots dans différents documents

Nous allons comparer des documents en fonction de la fréquence de mots dans chacun d'eux.

On charge d'abord le corpus Brown qui contient des extraits de journaux et revues, et on met dans new_text la liste des mots rencontrés. On va alors compter le nombre d'occurrences des mots modaux.

```python
from nltk.corpus import brown
news_text = brown.words(categories="news")
fdist = nltk.FreqDist([w.lower() for w in news_text])
modals = ['can', 'could', 'may', 'might', 'must', 'will']
for m in modals:
    print(m + ":", fdist[m])
```

Faites la même chose avec la catégorie 'science_fiction' (au lieu de 'news') et avec les mots commençant par wh (what, where, who, why...).
Nous allons maintenant faire une comparaison systématique entre catégories sur les mots modaux.

```python
cfd = nltk.ConditionalFreqDist(
    (genre, word)
    for genre in brown.categories()
    for word in brown.words(categories=genre))
```

```python
genres = ['news', 'religion','hobbies', 'science_fiction', 'romance', 'humor']
cfd.tabulate(conditions=genres, samples=modals)
```

On remarque que le mot modal le plus fréquent dans la catégorie 'news' est `will` et dans la catégorie 'romance' est `could`

On peut plus ou moins distinguer les catégories avec la fréquence des mots puisqu'elle permet d'identifier des tendances stylistiqes, de faire de la classification automatique de textes. Toutefois, cette métrique n'est pas suffisante seule pour faire la classification

##### 1.3. Les stopwords

On va d'abord examiner quels sont les mots les moins fréquents dans *Sense and Sensibility*. Pour cela, on calcule les mots présents dans ce roman et on en **retire tous les mots usuels en anglais**. Ceux-ci sont obtenus grâce à un corpus spécial de nltk. Voici donc la fonction qui fait ce travail: 

```python
def unusual_words(text):
    text_vocab = set(w.lower() for w in text if w.isalpha())
    english_vocab = set(w.lower() for w in nltk.corpus.words.words())
    unusual = text_vocab.difference(english_vocab)
    return sorted(unusual)
```

Ce qui donne:

```python
unusual_words(nltk.corpus.gutenberg.words('austen-sense.txt'))
```

Essayons sur le corpus de 'chat':

```python
unusual_words(nltk.corpus.nps_chat.words())
```

Il existe un **corpus de stopwords** disponible dans nltk. Il comporte 127 mots.

```python
from nltk.corpus import stopwords
stopwords.words('english')
```

Nous allons définir une fonction qui calcule la fraction de mots d'un texte qui ne font pas partie des stopwords. Nous l'appliquons alors au corpus des dépêches Reuters.

```python
def content_fraction(text):
    stopwords = nltk.corpus.stopwords.words('english')
    content = [w for w in text if w.lower() not in stopwords]
    return len(content) / len(text)
```

```python
content_fraction(nltk.corpus.reuters.words())
```

L'existence de WordNet permet de calculer des distances sémantiques entre mots. Par exemple, on peut considérer que deux mots sont d'autant plus proches sémantiquement qu'ils partagent un nyponyme à distance faible au-dessus d'eux.

Cherchons par exemple la similarité entre les mots 'right_whale', 'orca', 'minke', 'tortoise' et 'novel'.
On commence par charger le premier sense de chacun de ces mots:

```python
right = wn.synset('right_whale.n.01')
orca = wn.synset('orca.n.01')
minke = wn.synset('minke_whale.n.01')
tortoise = wn.synset('tortoise.n.01')
novel = wn.synset('novel.n.01')
```

On peut ensuite chercher l'hypernyme commun le plus bas dans l'ontologie:

```python
print(right.lowest_common_hypernyms(orca))
print(right.lowest_common_hypernyms(minke))
print(right.lowest_common_hypernyms(tortoise))
print(right.lowest_common_hypernyms(novel))
```

Il est claire que ‘right_whale’ et ‘whale’ sont proches, ce qui n’est pas le cas de ‘right_whale’ et de
‘novel’ dont l’hypernyme commun le plus spécifique est ‘entity’ qui est la racine de WordNet. On peut
quantifier cette proximité en regardant la profondeur du premier hypernyme commun.

```python
wn.synset('baleen_whale.n.01').min_depth()
```

```python
wn.synset('whale.n.02').min_depth()
```

```python
wn.synset('vertebrate.n.01').min_depth()
```

```python
wn.synset('entity.n.01').min_depth()
```

Une mesure découlant de cette « distance arborée » est calculée par la fonction path_similarity. Elle
retourne un nombre entre 0 et 1, 0 indiquant une similarité nulle alors que 1 est la similarité entre un terme
et lui-même (la fonction retourne -1 si aucun chemin n’est trouvé entre les deux termes (e.g. entre un nom
et un verbe).

```python
right.path_similarity(minke)
```

```python
right.path_similarity(orca)
```

```python
right.path_similarity(tortoise)
```

```python
right.path_similarity(novel)
```
